Finally, we will use the Interquartile Range (IQR) method to identify and count any statistical anomalies or potential outliers within the dataset.

```python
#We will use the IQR method for this business case
#write a function to calculate IQR and print rows with values that fall outside that IQR

def count_iqr_outliers(df, column):
    # define q1
    q1 = df[column].quantile(0.25)
    # define q3
    q3 = df[column].quantile(0.75)
    # define iqr
    iqr = q3 - q1
    # define outlier thresholds
    l_threshold = q1 - 1.5 * iqr
    u_threshold = q3 + 1.5 * iqr
    # dount outliers
    outliers = (df[column] < l_threshold) | (df[column] > u_threshold)
    # Count the number of True values (outliers)
    return outliers.sum()
```

```python
#iterate over the numerical columns of the dataframe:
num_var = list(df.select_dtypes(include=['int64', 'float64']).columns)

for column in num_var:
    print(f'{column} : {count_iqr_outliers(df, column)}')
```

It looks like we have numerous outliers across all three numerical variables. This result is somewhat expected because the dataset includes companies of vastly different sizes. As a result, companies with exceptionally large workforces and layoff events, such as Netflix, appear as statistical outliers.

These observations likely represent legitimate business events rather than data errors and should be retained for analysis, as they provide valuable insight into large-scale layoffs.

## Cleaning the Data

### Change Date Column to Datetime Format

Now that we have completed the data inspection process, we can begin cleaning the dataset.

The first step will be to convert the **date** column into a datetime format. This will ensure the column is properly structured for future analysis and allow us to perform date-based operations more efficiently.

```python
# Change date column type to datetime
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Verify change have been made
df.dtypes
```

### Removing Duplicates

Next, we will remove the five duplicate rows identified during the data inspection process. Eliminating these duplicate records will ensure that each entry in the dataset is unique and prevent duplicate information from impacting our analysis.

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Verify duplicates were removed
duplicates = df.duplicated().sum()
print(f"Duplicate Rows Remaining: {duplicates}")
```

### Removing Leading/Trailing Spaces

Now, we will remove the leading and trailing spaces identified in the company column. Cleaning these formatting inconsistencies will help standardize company names and prevent issues when grouping, filtering, or analyzing the data.

```python
# Remove leading and trailing spaces from the company column
df.loc[:, "company"] = df["company"].str.strip()
```

```python
# Verify no leading/trailing spaces remain in the company column
spaces = (
    df["company"]
    .dropna()
    .astype(str)
    .str.startswith(" ")
    |
    df["company"]
    .dropna()
    .astype(str)
    .str.endswith(" ")
).sum()

print(f"Leading/Trailing Spaces Remaining: {spaces}")
```

Earlier, we identified three different naming conventions for the Crypto industry category and two different naming conventions for the United States country category.

To standardize these values, we will update any industry values that begin with "Crypto" to "Crypto" and any country values that begin with "United States" to "United States". This will consolidate duplicate categories and ensure consistency throughout the dataset.

```python
# Standardize industry names to "Crypto"
df.loc[df["industry"].str.startswith("Crypto", na=False), "industry"] = "Crypto"

# Verify the update
print(sorted(df["industry"].dropna().unique()))
```

```python
# Standardize country names to "United States"
df.loc[df["country"].str.startswith("United States", na=False), "country"] = "United States"

# Verify the update
print(sorted(df["country"].dropna().unique()))
```

### Removing Null Values

Earlier we also found that our dataset had many null values in the **percentage_laid_off**, **total_laid_off**, and **total_funds_raised** columns. Since there are so many null values and there isn't an quick/easy way fill the data, we'll leave these as null for now.  


However, let's address the missing values in the **industry** column. As identified earlier, there are four rows where the industry value is null. We will examine these records to determine whether the same companies appear elsewhere in the dataset. If a company has multiple entries and one record contains an industry value while another is missing it, we can use the available information to update the null value and ensure consistency across the dataset.

```python
# Get the companies with missing industry values
missing_companies = df.loc[df["industry"].isna(), "company"].unique()

# Display all records for each company
for company in missing_companies:
    print(f"Company: {company}")
    print(df[df["company"] == company])
```

The companies missing an industry value are **Airbnb**, **Bally's Interactive**, **Juul**, and **Carvana**.

The good news is that Airbnb, Juul, and Carvana appear multiple times throughout the dataset, allowing us to use the industry values from their other records to fill in the missing entries and maintain consistency.

After researching Bally's Interactive, I determined that the company best fits within the Other industry category.

With this information, we can now update the dataset and fill in the missing values in the industry column.

```python
# Update industry values for specific companies

df.loc[df["company"] == "Airbnb", "industry"] = "Travel"

df.loc[df["company"] == "Juul", "industry"] = "Consumer"

df.loc[df["company"] == "Carvana", "industry"] = "Transportation"

df.loc[df["company"] == "Bally's Interactive", "industry"] = "Other"
```
